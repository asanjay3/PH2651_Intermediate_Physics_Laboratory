```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# =========================
# INPUT YOUR DATA HERE
# =========================
# Format: [Parallel Plate Voltage (V), Coil Current (A), Deflection (cm)]
data = np.array([
    [0.0, 0.0, 0.0],
    [500.0, 0.0, 0.35],
    [1000.0, 0.0, 1.3],
    [0.0, 0.47, -1.01],
    [500.0, 0.47, -0.38],
    [1000.0, 0.47, 0.3],
    [0.0, 0.92, -2.03],
    [500.0, 0.92, -1.41],
    [1000.0, 0.92, -0.59],
    # ... add all 30 rows here ...
])

# Split into x, y, z
x = data[:, 0]  # PP Voltage (V)
y = data[:, 1]  # Coil Current (A)
z = data[:, 2]  # Deflection (cm)

# =========================
# CREATE 3D PLOT
# =========================
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Scatter plot
#ax.scatter(x, y, z)

sc = ax.scatter(x, y, z, 
                s=80,                # size (increase this)
                c=z,                 # color by deflection
                cmap='viridis',      # color map
                edgecolors='black',
                depthshade=True)

plt.colorbar(sc, label='Deflection (cm)')

# Labels
ax.set_xlabel('PP Voltage (V)')
ax.set_ylabel('Coil Current (A)')
ax.set_zlabel('Deflection (cm)')

# Title
ax.set_title('Electron Deflection due to Varying Electric and Magnetic Field')
ax.view_init(elev=25, azim=45)


plt.show()
```
